```python
# Run to allow Collab to GitHub pushes
from google.colab import userdata

try:
    GITHUB_TOKEN = userdata.get('GITHUB_TOKEN')
    print("✓ GitHub token loaded")
except userdata.SecretNotFoundError:
    raise RuntimeError("Set GITHUB_TOKEN via the key icon in the left sidebar")
```

```python
import os
import subprocess

REPO_OWNER = "TestOrganizationLPRC"
REPO_NAME = "TestRepo1"
REPO_DIR = f"/content/{REPO_NAME}"

repo_url = f"https://{GITHUB_TOKEN}@github.com/{REPO_OWNER}/{REPO_NAME}.git"

if os.path.exists(REPO_DIR):
    subprocess.run(["git", "-C", REPO_DIR, "pull"], check=True)
else:
    result = subprocess.run(
        ["git", "clone", repo_url, REPO_DIR],
        capture_output=True, text=True
    )
    if result.returncode != 0:
        safe = result.stderr.replace(GITHUB_TOKEN, "***")
        raise RuntimeError(f"Clone failed:\n{safe}")
print("✓ Repo ready")
```

```python
# Requirements Install
result = subprocess.run(
    ["pip", "install", "-q", "-r", f"{REPO_DIR}/requirements.txt"],
    capture_output=True, text=True
)
if result.returncode != 0:
    print(result.stdout)
    print(result.stderr)
    raise RuntimeError("pip install failed")
print("✓ Dependencies installed")
```
